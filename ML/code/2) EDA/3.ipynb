# For Numerical - Numerical (Scatterplot)

# for Numerical - Catogorical (Bar plot)

```python
sns.barplot(x=titanic['Pclass'], y=titanic['Fare'])
```

## multivariant

```python
sns.barplot(x=titanic['Pclass'], y=titanic['Fare'],
            hue=titanic['Sex'],
            )
```

# for Numerical - Categorical (box-plot)

```python
sns.boxplot(x=titanic['Sex'], y=titanic['Age'],
            hue=titanic['2urvived']
            )
```

# for Numerical - Categorical (dist-plot)

```python
sns.displot(titanic[titanic['2urvived']==0]['Age'],kde=True)
sns.displot(titanic[titanic['2urvived']==1]['Age'],kde=True)
```

# for Catagorical - Catagorical (Heatmap)

```python
pd.crosstab(titanic['Pclass'],titanic['2urvived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['2urvived']))
```

```python
titanic.groupby('Pclass').mean()['2urvived']*100
```

# for Catagorical - Catagorical (Clustermap)

```python
pd.crosstab(titanic['sibsp'],titanic['2urvived'])
```

```python
sns.clustermap(pd.crosstab(titanic['sibsp'],titanic['2urvived']))
```
