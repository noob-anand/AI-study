# EDA using Univariate analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = pd.read_csv(r"E:\code\AI study\ML\code\2) EDA\train_and_test2.csv")
```

```python
df.sample(5)
```

# 1 Categorical Data

## a. Countplot

```python
print(df['2urvived'].value_counts())
df['2urvived'].value_counts().plot(kind='bar', title='Count of Survived and Not Survived')
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
print(df['Pclass'].value_counts())
df['Pclass'].value_counts().plot(kind='bar', title='people travelling in class')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

```python
df['Sex'].value_counts().plot(kind='bar', title='people travelling in class')
```

## b. piechart

```python
print(df['Pclass'].value_counts())
df['Pclass'].value_counts().plot(kind='pie', title='people travelling in class',autopct='%.2f%%')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.show()
```

# 2. Numerical Data

## a. Histogram

```python
plt.hist(df['Age'],bins=10)
```

## b. Displot

```python
sns.displot(df['Age'],kde=True,bins=10) # Better then histogram
```

## c. Boxplot

```python
sns.boxplot(x=df['Age']) # Boxplot is better than histogram and displot
```
