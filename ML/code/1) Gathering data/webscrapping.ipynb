```python
import pandas as pd
import requests
from bs4 import BeautifulSoup # for web scraping
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36",
    "Accept": "text/html,application/xhtml+xml,application/xml;q=0.9,image/avif,image/webp,image/apng,*/*;q=0.8",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept-Encoding": "gzip, deflate, br",
    "Connection": "keep-alive",
    "Upgrade-Insecure-Requests": "1",
}
webpage = requests.get("https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page=2", headers=headers).text
```

```python

soup = BeautifulSoup(webpage, "html.parser")
```

```python
company=soup.find_all('div', class_='companyCardWrapper')
len(company)
```

```python
name=[]
rating=[]
job=[]
reviews=[]
salary=[]
jobs=[]

for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find('div', class_='rating_text rating_text--md').text.strip())
    job.append(i.find('span', class_='companyCardWrapper__interLinking').text.strip())
    reviews.append(i.find('span', class_='companyCardWrapper__companyRatingCount').text.strip())
    salary.append(i.find_all('a', class_='companyCardWrapper__ActionWrapper')[1].text.strip())
    jobs.append(i.find_all('a', class_='companyCardWrapper__ActionWrapper')[3].text.strip())
```

```python
d={
    "Company Name": name,
    "Rating": rating,
    "Job Openings": job,
    "salary": salary,
    "jobs": jobs,
    "Reviews": reviews
}
df=pd.DataFrame(d)
df
```
