# to convert thingsd while loading

```python
def rename(name):
    if name == "Delhi Daredevils":
        return "DC"
    if name == "Mumbai Indians":
        return "MI"
    if name == "Royal Challengers Bangalore":
        return "RCB"
    if name == "Kolkata Knight Riders":
        return "KKR"
    if name == "Kings XI Punjab":
        return "KXIP"
    if name == "Rajasthan Royals":
        return "RR"
    if name == "Sunrisers Hyderabad":
        return "SRH"
    else:
        return name
```

```python
url = "https://raw.githubusercontent.com/12345k/IPL-Dataset/refs/heads/master/IPL/data.csv"
c = pd.read_csv(url,
                converters={'team1':rename, 'team2':rename})
c
```
