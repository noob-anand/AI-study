```python
df=df.iloc[:,2:]
df.head()
```

```python
from sklearn.model_selection import train_test_split

xtrain,xtest,ytrain,ytest=train_test_split(df.iloc[:,:2],
                                           df.iloc[:,-1],
                                           test_size=0.2
                                           )

xtrain
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe=OrdinalEncoder(categories=[['Poor','Average','Good'],['School','UG','PG']])

oe.fit(xtrain)

xtrainscaled=oe.transform(xtrain)
xtestscaled=oe.transform(xtest)
```

```python
xtrainscaled = pd.DataFrame(xtrainscaled, columns=xtrain.columns)
xtestscaled = pd.DataFrame(xtestscaled, columns=xtest.columns)

xtrainscaled
```
