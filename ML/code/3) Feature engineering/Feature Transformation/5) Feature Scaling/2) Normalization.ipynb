# Normalization using min-max scaling

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"E:\code\AI study\ML\code\3) Feature engineering\Social_Network_Ads.csv")
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split

xtrain,xtest,ytrain,ytest=train_test_split(df.drop('Purchased',axis=1),
                                           df['Purchased'],
                                           test_size=0.3,
                                           random_state=0
                                           )
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaler.fit(xtrain)

xtrainscaled=scaler.transform(xtrain)
xtestscaled=scaler.transform(xtest)

xtrainscaled
```

# changing them from numpy array to dataframe

```python
xtrainscaled = pd.DataFrame(xtrainscaled, columns=xtrain.columns)
xtestscaled = pd.DataFrame(xtestscaled, columns=xtest.columns)

xtrainscaled
```
